```python
stu_name = input("Enter name of the student: ")
stu_reg_id = int(input("Enter reg number: "))
telugu = int(input("Enter marks of Telugu subject: "))
hindi = int(input("Enter marks of Hindi subject: "))
english = int(input("Enter marks of English subject: "))
maths = int(input("Enter marks of Maths subject: "))
science = int(input("Enter marks of Science subject: "))
social = int(input("Enter marks of Social subject: "))

total = telugu + hindi + english + maths + science + social
average_marks = total / 6

if (telugu >= 35 and hindi >= 35 and english >= 35 and
    maths >= 35 and science 
    >= 40 and social > 35):

    print("PASS")

    if average_marks >= 75:
        grade = "Distinction"
    elif average_marks >= 60:
        grade = "First Class"
    elif average_marks >= 50:
        grade = "Second Class"
    elif average_marks >= 35:
        grade = "Third Class"
    else:
        grade = "No Grade"

    print("Grade:", grade)

else:
    print("FAIL")

print("Total Marks:", total)
print("Average Marks:", average_marks)
```
